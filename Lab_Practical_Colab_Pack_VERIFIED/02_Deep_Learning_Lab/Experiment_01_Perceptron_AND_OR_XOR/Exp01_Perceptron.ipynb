# Experiment 1 - Single Unit Perceptron

Train a perceptron on AND, OR and XOR truth tables. XOR is not linearly separable, so a single perceptron cannot learn it perfectly.

**Run cells from top to bottom.** Installation is included where required.

```python
%pip -q --disable-pip-version-check install numpy scikit-learn
```

```python
import numpy as np
from sklearn.linear_model import Perceptron
from sklearn.metrics import accuracy_score

X = np.array([[0,0],[0,1],[1,0],[1,1]])
gates = {"AND":[0,0,0,1], "OR":[0,1,1,1], "XOR":[0,1,1,0]}
for name, values in gates.items():
    y = np.array(values)
    model = Perceptron(max_iter=1000, random_state=0, tol=1e-3)
    model.fit(X, y)
    pred = model.predict(X)
    print("\\n", name, "Actual:", y, "Predicted:", pred)
    print("Accuracy:", accuracy_score(y, pred))
```
