# Experiment 2 - ANN Activation Functions

Compare sigmoid, tanh and ReLU on MNIST. One epoch per model keeps the practical run relatively short.

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten, Dense

(X_train, y_train), (X_test, y_test) = mnist.load_data()
X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0
```

```python
from tensorflow.keras.layers import Input
results = {}
for act in ["sigmoid", "tanh", "relu"]:
    model = Sequential([
        Input(shape=(28,28)),
        Flatten(),
        Dense(128, activation=act),
        Dense(64, activation=act),
        Dense(10, activation="softmax")
    ])
    model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
    model.fit(X_train, y_train, epochs=1, batch_size=128, verbose=0)
    loss, acc = model.evaluate(X_test, y_test, verbose=0)
    results[act] = round(float(acc), 4)
print("Test accuracy by activation:", results)
```
