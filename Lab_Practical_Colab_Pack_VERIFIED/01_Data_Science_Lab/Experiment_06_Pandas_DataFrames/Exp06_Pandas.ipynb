```python
import pandas as pd

df = pd.DataFrame({
    "Name":["Arun","Bala","Cathy","Deepa"],
    "Marks":[85,90,78,95],
    "Age":[20,21,20,22]
})
display(df)
print("First rows:"); display(df.head())
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
print("Description:"); display(df.describe())
print("Mean marks:", df["Marks"].mean())
print("Marks > 80:"); display(df[df["Marks"] > 80])
print("Sorted by marks:"); display(df.sort_values("Marks"))
print("Missing values:"); print(df.isnull().sum())
```
