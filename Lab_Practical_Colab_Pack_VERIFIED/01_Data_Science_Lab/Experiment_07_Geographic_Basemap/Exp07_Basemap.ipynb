```python
import matplotlib.pyplot as plt
try:
    from mpl_toolkits.basemap import Basemap
    fig, ax = plt.subplots(figsize=(7, 5))
    m = Basemap(projection="merc", llcrnrlat=5, urcrnrlat=35,
                llcrnrlon=65, urcrnrlon=100, resolution="c", ax=ax)
    m.drawcoastlines()
    m.drawcountries()
    m.drawmapboundary()
    lon, lat = m(80.2707, 13.0827)  # Chennai
    m.plot(lon, lat, "ro", markersize=6)
    ax.set_title("Geographic Data - Chennai")
    plt.show()
except Exception as e:
    print("Basemap unavailable:", e)
    print("Using simple longitude/latitude plot instead.")
    plt.close("all")
    fig, ax = plt.subplots(figsize=(7, 5))
    ax.scatter([80.2707], [13.0827], s=30)
    ax.text(80.2707, 13.0827, " Chennai")
    ax.set_xlabel("Longitude")
    ax.set_ylabel("Latitude")
    ax.set_title("Geographic Data - Chennai")
    ax.grid(True)
    plt.show()
```
