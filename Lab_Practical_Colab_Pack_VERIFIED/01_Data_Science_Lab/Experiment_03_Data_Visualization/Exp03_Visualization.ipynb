```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
from scipy.stats import norm

df = load_iris(as_frame=True).frame
x = df["sepal length (cm)"]
y = df["sepal width (cm)"]
z = df["petal length (cm)"]

# 1 Normal curve
xx = np.linspace(x.min(), x.max(), 100)
plt.plot(xx, norm.pdf(xx, x.mean(), x.std()))
plt.title("Normal Curve"); plt.xlabel("Sepal length"); plt.ylabel("Density"); plt.show()

# 2 Density plot
sns.kdeplot(x=x, y=y, fill=True)
plt.title("Density Plot"); plt.show()

# 3 Contour plot
plt.tricontour(x, y, z, levels=10)
plt.colorbar(label="Petal length")
plt.title("Contour Plot"); plt.xlabel("Sepal length"); plt.ylabel("Sepal width"); plt.show()

# 4 Correlation heatmap
sns.heatmap(df.iloc[:, :4].corr(), annot=True)
plt.title("Correlation Heatmap"); plt.show()

# 5 Scatter plot
plt.scatter(x, z)
plt.xlabel("Sepal length"); plt.ylabel("Petal length"); plt.title("Scatter Plot"); plt.show()

# 6 Histogram
plt.hist(x, bins=10)
plt.xlabel("Sepal length"); plt.ylabel("Frequency"); plt.title("Histogram"); plt.show()

# 7 Three-dimensional plot
fig = plt.figure()
ax = fig.add_subplot(111, projection="3d")
ax.scatter(df.iloc[:,0], df.iloc[:,1], df.iloc[:,2])
ax.set_xlabel("Sepal length"); ax.set_ylabel("Sepal width"); ax.set_zlabel("Petal length")
plt.title("3D Plot"); plt.show()
```
