```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "City":["Chennai","Bangalore","Mumbai","Delhi"],
    "Latitude":[13.08,12.97,19.07,28.61],
    "Longitude":[80.27,77.59,72.87,77.20]
})
fig, ax = plt.subplots(figsize=(7, 5))
ax.scatter(df["Longitude"], df["Latitude"], s=30)
for _, row in df.iterrows():
    ax.text(row["Longitude"], row["Latitude"], " " + row["City"])
ax.set_xlabel("Longitude")
ax.set_ylabel("Latitude")
ax.set_title("Geographic Data")
ax.grid(True)
plt.show()
```
