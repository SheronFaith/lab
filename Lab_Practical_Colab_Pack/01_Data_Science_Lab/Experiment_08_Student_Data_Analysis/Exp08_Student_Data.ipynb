```python
import pandas as pd
df = pd.DataFrame({
    "Name":["A","B","C","D"],
    "Marks":[85,90,75,95],
    "Age":[20,21,20,22]
})
display(df)
print("Average:", df["Marks"].mean())
print("Highest:", df["Marks"].max())
print("Lowest:", df["Marks"].min())
display(df[df["Marks"] >= 80])
```
