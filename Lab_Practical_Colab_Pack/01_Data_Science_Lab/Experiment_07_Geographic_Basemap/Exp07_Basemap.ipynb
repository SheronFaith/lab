```python
import matplotlib.pyplot as plt
try:
    from mpl_toolkits.basemap import Basemap
    m = Basemap(projection="merc", llcrnrlat=5, urcrnrlat=35,
                llcrnrlon=65, urcrnrlon=100, resolution="c")
    m.drawcoastlines()
    m.drawcountries()
    m.drawmapboundary()
    lon, lat = m(80.2707, 13.0827)  # Chennai
    m.plot(lon, lat, "ro")
    plt.title("Geographic Data - Chennai")
    plt.show()
except Exception as e:
    print("Basemap unavailable:", e)
    print("Using simple longitude/latitude plot instead.")
    plt.scatter([80.2707], [13.0827])
    plt.text(80.2707, 13.0827, "Chennai")
    plt.xlabel("Longitude"); plt.ylabel("Latitude")
    plt.title("Geographic Data"); plt.grid(True); plt.show()
```
