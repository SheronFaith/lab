```python
import numpy as np

a = np.array([10,20,30,40])
b = np.array([1,2,3,4])
print("Array:", a)
print("Shape:", a.shape)
print("Dimensions:", a.ndim)
print("Addition:", a+b)
print("Subtraction:", a-b)
print("Multiplication:", a*b)
print("Division:", a/b)
print("Mean:", np.mean(a))
print("Max:", np.max(a))
print("Min:", np.min(a))
m = np.array([[1,2],[3,4]])
print("Matrix:\\n", m)
print("Transpose:\\n", m.T)
print("Reshape:", np.arange(6).reshape(2,3))
```
