```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "City":["Chennai","Bangalore","Mumbai","Delhi"],
    "Latitude":[13.08,12.97,19.07,28.61],
    "Longitude":[80.27,77.59,72.87,77.20]
})
plt.scatter(df["Longitude"], df["Latitude"])
for _, row in df.iterrows():
    plt.text(row["Longitude"], row["Latitude"], row["City"])
plt.xlabel("Longitude"); plt.ylabel("Latitude")
plt.title("Geographic Data"); plt.grid(True); plt.show()
```
