# Experiment 9 - Data Augmentation from a Single Image

Generate rotated, zoomed and flipped variants of an image. Uses a synthetic sample so no image upload is needed.

```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.preprocessing.image import ImageDataGenerator

image = np.random.rand(1,100,100,3).astype("float32")
datagen = ImageDataGenerator(rotation_range=30, horizontal_flip=True, zoom_range=0.2)
plt.figure(figsize=(8,3))
for i, batch in enumerate(datagen.flow(image,batch_size=1), start=1):
    plt.subplot(1,4,i)
    plt.imshow(batch[0])
    plt.axis("off")
    if i == 4:
        break
plt.suptitle("Augmented Image Samples")
plt.show()
```
